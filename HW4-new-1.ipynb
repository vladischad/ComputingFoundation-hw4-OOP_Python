**Q1.  (10 pts) Angle between Stars**. Given two stars with angles of
    declination and right ascension ($d_1$, $a_1$) and ($d_2$, $a_2$),
    respectively, the angle they subtend is given by the formula
    $$2 \arcsin\left\{[\sin^2(d/2) + \cos(d_1)\cos(d_2)\sin^2(a/2)]^{1/2}\right\},$$
    where $a_1$ and $a_2$ are angles between -180 and 180 degrees, $d_1$
    and $d_2$ are angles between -90 and 90 degrees, $a = a_2-a_1$, and
    $d = d_2 - d_1$. Compose a program to take the declination and right
    ascension of two stars as command-line arguments and write the angle
    they subtend. Select the coordinates of the following two bright
    stars.
    **Hint**: Be careful about converting from degrees to radians.

  | Star’s name |      Right  Ascension  (RA) | Declination  (DEC)    |
  |-------------|-----------------------------|-----------------------|
  | Altair      | 19h  50m    46.99855s       | +08&deg;52’          05.9563” |
  | Vega        | 18h  36m    56.33635s       | +38&deg;47’          01.2802” |
  
       
Here the right ascension is represented using "hour angles\", so you need to convert it to degrees first. Basically, 3600 s = 60 m = 1 h = 15$^\circ$. For example, 23h=345$^\circ$=-15$^\circ$. 

**Extended reading.** Any units of angular measure could have been chosen for right ascension, but it is customarily measured in hours (h), minutes (m), and seconds (s), with 24h being equivalent to a full circle (360 $^\circ$). Astronomers have chosen this unit to measure right ascension because they measure a star's location by timing its passage through the highest point in the sky as the Earth rotates. The line which passes through the highest point in the sky, called the meridian, is the projection of a longitude line onto the celestial sphere. Since a complete circle contains 24h of right ascension or 360$^\circ$ (degrees of arc), 1/24 of a circle is measured as 1h of right ascension, or 15$^\circ$; 1/1440 of a circle is measured as 1m of right ascension, or 15 minutes of arc (also written as 15'); and 1/86400 of a circle contains 1s of right ascension, or 15 seconds of arc (also written as 15\"). A full circle, measured in right-ascension units, contains $24\times60\times60=86400$s or $24\times60=1440$m, or 24h.

````python
def ra_to_deg(h, m, s):
    # convert everything to hours first
    deg = (h + m/60 + s/3600) * 15
    if deg > 180:
        deg -= 360
    return deg

def dec_to_deg(deg, arcmin, arcsec):
    return deg + arcmin/60 + arcsec/3600

def star_angle(d1, a1, d2, a2):
    # convert degrees to radians
    d1, a1, d2, a2 = map(math.radians, [d1, a1, d2, a2])
    d = d2 - d1
    a = a2 - a1
    angle = 2 * math.asin(math.sqrt(math.sin(d/2)**2 + math.cos(d1)*math.cos(d2)*math.sin(a/2)**2))
    return math.degrees(angle)

a1 = ra_to_deg(19, 50, 46.99855)
d1 = dec_to_deg(8, 52, 5.9563)
a2 = ra_to_deg(18, 36, 56.33635)
d2 = dec_to_deg(38, 47, 1.2802)

print(f"Altair: RA = {a1:.6f} deg, DEC = {d1:.6f} deg")
print(f"Vega:   RA = {a2:.6f} deg, DEC = {d2:.6f} deg")
print(f"Angle between Altair and Vega = {star_angle(d1, a1, d2, a2):.4f} degrees")
````

**Q2. (10 pts) Matrix multiplication.** Compose a function multiply() in a Matrix class
    that takes two square matrices of the same dimension and returns
    their product. Make sure your output is formatted.

$$
    \begin{pmatrix} 
        1       & 2      & 3  \\

        7       & 8      & 9 \\
    \end{pmatrix}
    \cdot
    \begin{pmatrix}

        3       & 2      & 1 \\
    \end{pmatrix}
    =?
$$

**Extended Reading.** If you are not familiar with the matrix
    multiplication, check your math book for details. We briefly
    summarize it below. If you have 2 matrices **A** and **B**,
    their product **C** is represented as, 

    
$$
\begin{matrix} 
C &= A \cdot B \\
A &=[a_{ij}] \\
B &=[b_{ij}] \\
C &=[c_{ij}] \\
\end{matrix}
$$
    
The matrix forms are, 
    
$$
    \begin{pmatrix} 
        c_{11}       & c_{12}  & \dots & c_{1n} \\
        c_{21}       & c_{22} & \dots & c_{2n} \\
        \vdots       & \vdots   & \ddots & \vdots\\
        c_{n1}       & c_{n2}  & \dots & c_{nn}
        \end{pmatrix}
    =
    \begin{pmatrix} 
        a_{11}       & a_{12}  & \dots & a_{1n} \\
        a_{21}       & a_{22}  & \dots & a_{2n} \\
        \vdots       & \vdots   & \ddots & \vdots\\
        a_{n1}       & a_{n2}  & \dots & a_{nn}
    \end{pmatrix}
    \cdot
    \begin{pmatrix} 
        b_{11}       & b_{12}  & \dots & b_{1n} \\
        b_{21}       & b_{22}  & \dots & b_{2n} \\
        \vdots       & \vdots   & \ddots & \vdots\\
        b_{n1}       & b_{n2}  & \dots & b_{nn}
    \end{pmatrix}
$$ 
    
Each term in **C** is computed as,

$$c_{ij}=\sum_{k=1}^{n}a_{ik}\cdot b_{kj}$$


````python
class Matrix:
    def __init__(self, data):
        self.data = data
        self.n = len(data)

    def multiply(self, other):
        n = self.n
        result = [[0] * n for _ in range(n)]
        for i in range(n):
            for j in range(n):
                for k in range(n):
                    result[i][j] += self.data[i][k] * other.data[k][j]
        return Matrix(result)

    def __str__(self):
        rows = []
        for row in self.data:
            rows.append(" ".join(f"{x:5d}" for x in row))
        return "\n".join(rows)

A = Matrix([[1, 2, 3], [4, 5, 6], [7, 8, 9]])
B = Matrix([[9, 8, 7], [6, 5, 4], [3, 2, 1]])
C = A.multiply(B)

print("A =")
print(A)
print("B =")
print(B)
print("A * B =")
print(C)
````

**Q3. (10pts) Polynomials using array**. Compose a function `evaluate(x, a)` that evaluates the
    polynomial $a(x)$ whose coefficients are the elements in the array
    $a[]$: $$a_0+a_1x+a_2x^2+...+a_{n-2}x^{n-2}+a_{n-1}x^{n-1}$$ An
    efficient way to perform the computations that is suggested by the
    following parenthesization:
    $$a_0+x(a_1+ x(a_2+...+x(a_{n-2}+xa_{n-1})...))$$ Then compose a
    function exp() that calls evaluate() to compute an approximation to
    $e^x$, using the first $n$ terms of the Taylor series expansion
    $e^x = 1 + x + x^2/2! + x^3/3! + ....$. Compute $e^1=?$ as an example, and compare your result againstthat computed by `math.exp(x)`.

````python
def evaluate(x, a):
    # start from the last coefficient
    result = 0
    for coef in reversed(a):
        result = coef + x * result
    return result

def exp(x, n=20):
    # coefficients of the Taylor series are 1/k!
    a = [1 / math.factorial(k) for k in range(n)]
    return evaluate(x, a)

x = 1
mine = exp(x)
real = math.exp(x)
print(f"My exp(1)   = {mine:.15f}")
print(f"math.exp(1) = {real:.15f}")
print(f"Difference  = {abs(mine - real):.2e}")

for n in [2, 5, 10, 15]:
    print(f"n = {n:2d}: {exp(x, n):.15f}")
````

**Days of the week.**

Compose a program that accepts a date as input and writes the day of the week that date falls on. 

- There should be 3 variables: m (month), d (day), and y (year). 
- For m, use 1 for January, 2 for February, and so forth.
- For days of week, use 0 for Sunday, 1 for Monday, 2 for Tuesday, and so forth.

We can use the following formulas for the Gregorian calendar to compute the days of the week denoted as **d0**:

```
y0 = y − (14 − m)//12
x = y0 + y0//4 − y0//100 + y0//400
m0 = m + 12 × ((14 − m)//12)− 2
d0 = (d + x + (31 × m0)//12)%7
```

where “//” denotes floored division operator and “%” denotes the remainder operator. 

For example, let's computer what day of the week was February 14, 2000?
```
y0 = 2000 − (14 − 2)//12 = 1999
x = 1999 + 1999//4− 1999//100 + 1999//400 = 1999 + 499− 19 + 4 = 2483
m0 = 2 + 12 × ((14 − 2)//12) − 2 = 2 + 12 × 1 − 2 = 12
d0 = (14 + 2483 + (31 × 12)//12)%7 = (14 + 2483 + 31)%7 = 1(Monday)
```

So 2020/2/14 is Monday. 

````python
# Define an Enum for representing days of the week
class DayOfWeek(Enum):
    '''
    The left are names for Enum members and the right are values. 
    '''
    SUNDAY = 0
    MONDAY = 1
    TUESDAY = 2
    WEDNESDAY = 3
    THURSDAY = 4
    FRIDAY = 5
    SATURDAY = 6

#Let's try some basic usages of Enum. Please learn from these examples, we may need it later.

print(DayOfWeek.SUNDAY.name)
print(DayOfWeek.SUNDAY.value)
print(DayOfWeek(0).name)
````

````python
# Define a day_of_week function

def get_day_of_week(m, d, y):
    """
    Get the day of the week for a given date.

    Parameters:
    m (month), d (day), and y (year).

    Outputs:
    d0: the day of the week
    
    """

    y0 = y - (14 - m) // 12
    x = y0 + y0 // 4 - y0 // 100 + y0 // 400
    m0 = m + 12 * ((14 - m) // 12) - 2
    d0 = (d + x + (31 * m0) // 12) % 7
    
    return DayOfWeek(d0)
````

Instead, use the following simpler method: repeatedly divide 2 into
n and read the remainders backward. First, compose a while loop to
carry out this computation and write the bits in the wrong order.
Then, use **recursion** to write the bits in the correct order.


````python
# Part 1: while loop, bits come out in the wrong (reverse) order
def binary_reversed(n):
    while n > 0:
        print(n % 2, end="")
        n //= 2
    print()

# Part 2: recursion, prints the bits in the right order
def binary(n):
    if n > 1:
        binary(n // 2)
    print(n % 2, end="")

# In a .py file n would come from the command line: n = int(sys.argv[1])
for n in [8, 13, 37, 255]:
    print(f"n = {n}")
    print("  reversed: ", end="")
    binary_reversed(n)
    print("  correct:  ", end="")
    binary(n)
    print()
````
